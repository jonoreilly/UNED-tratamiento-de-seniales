## Captura de la señal

### Instalar dependencias

```julia
using Pkg
Pkg.add("WAV")
Pkg.add("Plots")
```

### Grabar audio

El fichero de audio que he capturado se encuentra en 'audio/ukelele.wav'

### Reproducir audio

```julia
using WAV
y, fs = wavread(raw".\audio\ukelele.wav")
wavplay(y, fs)

# y {Matrix[n, 2]<number>} - contiene los datos del archivo audio en dos canales (columna 1: canal izquierdo, columna 2: canal derecho)
# fs {number} - frecuencia de muestreo, en muestras por segundo (Hz)
```

### Calcular la duración de la señal

Basta con dividir la longitud de la señal `y` por la frecuencia de muestreo `fs`.

```julia
using WAV
y, fs = wavread(raw".\audio\ukelele.wav")
longitud = size(y)[1]
duracion = longitud / fs
println("Duración: $duracion segundos")
```

### Calcular el valor máximo de la señal

Para empezar, la amplitud se refiere al valor absoluto de una muestra, tanto positiva como negativa. Para ello podemos transformar la señal a todo positivo primero, y luego encontrar el máximo de ellos.

También tenemos que tener en cuenta que hay 2 canales de audio (izquierda y derecha). Podemos encontrar la amplitud máxima de cada uno por separado, o tratarlos como "mono" y calcular la amplitud máxima de su media.

```julia
using WAV
y, fs = wavread(raw".\audio\ukelele.wav")

canalIzquierdo = y[:,1]
canalDerecho = y[:,2]
mono = (canalIzquierdo .+ canalDerecho) ./ 2

maxIzquierda = maximum(abs.(canalIzquierdo))
maxDerecha = maximum(abs.(canalDerecho))
maxMono = maximum(abs.(mono))

println("Amplitudes máximas:
    izquierda: $maxIzquierda
    derecha: $maxDerecha
    mono: $maxMono
")
```

### Calcular el porcentaje del rango utilizado

El rango total disponible se refiere al rango de valores entre los valores mínimo y máximo posibles para una señal. El rango utilizado es el rango de valores entre los valores mínimo y máximo utilizados por la señal.

Como `wavread` siempre devuelve muestras en formato Float con rango -1.0 a 1.0, ese es el rango disponible.

Podeoms calcular el rango utilizado usando las funciones `maximum` y `minimum`.

Para comparar los rangos lo más facil es calcular sus distancias entre sus minimos y sus máximos (d = max - min). Luego podemos calcular su relación, dividiendo la distancia del rango utilizado por la del rango disponible (% = d-utilizado / d-disponible).

Dependiendo de qué queramos hacer con esta información, podemos calcular el rango utilizado por cada canal o en total entre todos los canales.

```julia
using WAV
y, fs = wavread(raw".\audio\ukelele.wav")

canalIzquierdo = y[:,1]
canalDerecho = y[:,2]
todosLosCanales = [y[:,1]; y[:,2]]

maxIzquierda = maximum(canalIzquierdo)
maxDerecha = maximum(canalDerecho)
maxAbsoluto = maximum(todosLosCanales)

minIzquierda = minimum(canalIzquierdo)
minDerecha = minimum(canalDerecho)
minAbsoluto = minimum(todosLosCanales)

rangoIzquierda = maxIzquierda - minIzquierda
rangoDerecha = maxDerecha - minDerecha
rangoAbsoluto = maxAbsoluto - minAbsoluto

rangoDisponible = 1.0 - (-1.0)

println("Porcentaje del rango utilizado:
    izquierda: $((rangoIzquierda / rangoDisponible) * 100)%
    derecha: $((rangoDerecha / rangoDisponible) * 100)%
    absoluto: $((rangoAbsoluto / rangoDisponible) * 100)%
")
```
